```python
import numpy as np
import healpy as hp

# thickness of the lensing plane in Mpc/h
thickness = 50 
### index of cosmology from 0 to 128
ic    = 0
### index of weak lensing kappa file from 1 to ~90
### comoving distance of the source plane is [indwl * thickness] Mpc/h
indwl = 50 # corresponding to 2500 Mpc/h, which is close to the comoving distance of z_s=1.0 

nside = 1024
### main directory
fdir  = '/home/cossim/yuyu22/chenzhao/csst/simulation/'
fbase = fdir + 'c%04d/output/kappas/'%(ic) # if z_s < 0.5 change to "/kappas-le0.5/"
if nside == 8192:
    fname = "kappa-nbox%d-thickness%d-8192.bin"%(indwl, thickness)
elif nside == 1024:
    fname = "kappa-nbox%d-thickness%d-downfrom8192-%d.bin"%(indwl, thickness, nside)
else:
    raise ValueError("nside should be either 1024 or 8192")

kappa = np.fromfile(fbase + fname, dtype=np.float32)
print("kappa shape:", kappa.shape)
print("should be:", hp.nside2npix(nside))  
```

```python
hp.mollview(kappa, title="Full-sky kappa map under Born approximation")
```

```python
hp.mollzoom(kappa, title="zoomed kappa map under Born approximation")
```
